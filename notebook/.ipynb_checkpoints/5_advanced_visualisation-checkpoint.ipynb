# 5. Advanced Visualisation & Pose Reconstruction

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
import seaborn as sns
import pandas as pd
from matplotlib.lines import Line2D
from sklearn.manifold import TSNE
from IPython.display import Video

SIGNS = ["Hello", "ThankYou", "Yes", "No", "Help", "Rest"]
d = np.load("processed.npz", allow_pickle=True)
X, y, groups = d["X"], d["y"], d["groups"]            # X: (N, 90, 162) head-relative
coord_cols = list(d["coord_cols"])
f = np.load("features.npz", allow_pickle=True)
X_tab = f["X_tab"]
SUBJECTS = sorted(set(groups))
print("X:", X.shape, "| subjects:", SUBJECTS)
```

```python
FINGERS = {
    "thumb":  ["thumbKnuckle", "thumbIntermediateBase", "thumbIntermediateTip", "thumbTip"],
    "index":  ["indexFingerMetacarpal", "indexFingerKnuckle", "indexFingerIntermediateBase", "indexFingerIntermediateTip", "indexFingerTip"],
    "middle": ["middleFingerMetacarpal", "middleFingerKnuckle", "middleFingerIntermediateBase", "middleFingerIntermediateTip", "middleFingerTip"],
    "ring":   ["ringFingerMetacarpal", "ringFingerKnuckle", "ringFingerIntermediateBase", "ringFingerIntermediateTip", "ringFingerTip"],
    "little": ["littleFingerMetacarpal", "littleFingerKnuckle", "littleFingerIntermediateBase", "littleFingerIntermediateTip", "littleFingerTip"],
}
BONES = []                       # list of (jointA, jointB)
for chain in FINGERS.values():
    BONES.append(("wrist", chain[0]))
    BONES += list(zip(chain[:-1], chain[1:]))
BONES.append(("wrist", "forearmWrist"))
BONES.append(("forearmWrist", "forearmArm"))

def joint_xyz(window, hand, joint):
    i = coord_cols.index(f"{hand}_{joint}_x")
    return window[:, i:i+3]                          # (T, 3)

def hand_present(window, hand):
    i = coord_cols.index(f"{hand}_wrist_x")
    return np.abs(window[:, i:i+3]).sum() > 1e-6
```

### A. Class prototype motion — mean speed profile per sign

```python
def frame_speed(window):
    pts = window.reshape(window.shape[0], -1, 3)      # (T, 54, 3)
    v = np.linalg.norm(np.diff(pts, axis=0), axis=2)  # (T-1, 54)
    return v.mean(axis=1)                             # (T-1,)

plt.figure(figsize=(9, 4))
for s in SIGNS:
    prof = np.stack([frame_speed(X[i]) for i in np.where(y == s)[0]]).mean(0)
    plt.plot(prof, label=s)
plt.title("Mean joint speed over time (per sign)"); plt.xlabel("frame"); plt.ylabel("m / frame"); plt.legend()
plt.tight_layout(); plt.show()
```

### B. Total motion energy per sign, split by subject

```python
rows = []
for i in range(len(X)):
    rows.append((groups[i], y[i], frame_speed(X[i]).sum()))
mdf = pd.DataFrame(rows, columns=["subject", "sign", "energy"])
plt.figure(figsize=(11, 4))
sns.boxplot(data=mdf, x="sign", y="energy", hue="subject", order=SIGNS)
plt.title("Total motion energy by sign and subject"); plt.tight_layout(); plt.show()
```

### C. Feature-space structure — t-SNE coloured by sign and by subject

```python
emb = TSNE(n_components=2, random_state=0, perplexity=30, init="pca").fit_transform(X_tab)
fig, ax = plt.subplots(1, 2, figsize=(13, 5))
sns.scatterplot(x=emb[:, 0], y=emb[:, 1], hue=y, hue_order=SIGNS, s=30, ax=ax[0])
ax[0].set_title("t-SNE by sign")
sns.scatterplot(x=emb[:, 0], y=emb[:, 1], hue=groups, s=30, ax=ax[1])
ax[1].set_title("t-SNE by subject")
plt.tight_layout(); plt.show()
```

### D. Class centroid distance — how confusable are the signs

```python
cent = np.stack([X_tab[y == s].mean(0) for s in SIGNS])
D = np.linalg.norm(cent[:, None, :] - cent[None, :, :], axis=2)
plt.figure(figsize=(5.5, 4.5))
sns.heatmap(D, annot=True, fmt=".1f", cmap="magma_r", xticklabels=SIGNS, yticklabels=SIGNS)
plt.title("Pairwise class-centroid distance (smaller = more confusable)"); plt.tight_layout(); plt.show()
```

### E. Pose reconstruction — one sign, hand pose across all subjects

```python
SIGN = "Hello"
FRAME = 45                                          # mid-window representative pose

def rep_window(sign, subject):
    idx = np.where((y == sign) & (groups == subject))[0]
    return X[idx[0]]

def draw_hands(ax, window, frame, color, label=None):
    first = True
    for hand in ("L", "R"):
        if not hand_present(window, hand):
            continue
        P = {j: joint_xyz(window, hand, j)[frame] for _, j in
             [("", "wrist"), ("", "forearmWrist"), ("", "forearmArm")] +
             [("", j) for ch in FINGERS.values() for j in ch]}
        for a, b in BONES:
            xa, xb = P[a], P[b]
            ax.plot([xa[0], xb[0]], [xa[1], xb[1]], [xa[2], xb[2]],
                    color=color, lw=1.5, label=(label if first else None))
            first = False
        pts = np.array(list(P.values()))
        ax.scatter(pts[:, 0], pts[:, 1], pts[:, 2], color=color, s=8)

palette = sns.color_palette("tab10", len(SUBJECTS))
fig = plt.figure(figsize=(8, 7)); ax = fig.add_subplot(111, projection="3d")
for c, s in zip(palette, SUBJECTS):
    draw_hands(ax, rep_window(SIGN, s), FRAME, c, label=s)
ax.set_title(f"'{SIGN}' hand pose across subjects (head-relative, frame {FRAME})")
ax.set_xlabel("x (m)"); ax.set_ylabel("y (m)"); ax.set_zlabel("z (m)")
ax.legend(handles=[Line2D([0], [0], color=c, label=s) for c, s in zip(palette, SUBJECTS)])
ax.view_init(elev=15, azim=-70); plt.tight_layout(); plt.show()
```

### F. Pose reconstruction video — the sign animated across all subjects

```python
windows = {s: rep_window(SIGN, s) for s in SUBJECTS}
allpts = np.concatenate([w.reshape(-1, 3) for w in windows.values()])
allpts = allpts[np.abs(allpts).sum(1) > 1e-6]
lo, hi = allpts.min(0), allpts.max(0)

fig = plt.figure(figsize=(7, 6)); ax = fig.add_subplot(111, projection="3d")
def update(t):
    ax.cla()
    for c, s in zip(palette, SUBJECTS):
        draw_hands(ax, windows[s], t, c, label=s)
    ax.set_xlim(lo[0], hi[0]); ax.set_ylim(lo[1], hi[1]); ax.set_zlim(lo[2], hi[2])
    ax.set_title(f"'{SIGN}'  frame {t}/{X.shape[1]-1}")
    ax.set_xlabel("x"); ax.set_ylabel("y"); ax.set_zlabel("z")
    ax.view_init(elev=15, azim=-70)
    ax.legend(handles=[Line2D([0], [0], color=c, label=s) for c, s in zip(palette, SUBJECTS)], loc="upper right")

anim = animation.FuncAnimation(fig, update, frames=X.shape[1], interval=50)
anim.save(f"pose_{SIGN}.mp4", writer=animation.FFMpegWriter(fps=20))
plt.close(fig)
print("saved", f"pose_{SIGN}.mp4")
Video(f"pose_{SIGN}.mp4", embed=True)
```
