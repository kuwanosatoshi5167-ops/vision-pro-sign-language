# 4. ML Development (LOSO)

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import torch
import torch.nn as nn
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import LeaveOneGroupOut
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

SIGNS = ["Hello", "ThankYou", "Yes", "No", "Help", "Rest"]
np.random.seed(0); torch.manual_seed(0)
d = np.load("features.npz", allow_pickle=True)
X_tab, X_seq, y_idx, groups = d["X_tab"], d["X_seq"], d["y_idx"], d["groups"]
logo = LeaveOneGroupOut()
```

```python
def report(true, pred, title):
    print(title, "accuracy:", round(accuracy_score(true, pred), 3))
    print(classification_report(true, pred, target_names=SIGNS, zero_division=0))
    cm = confusion_matrix(true, pred, labels=range(len(SIGNS)))
    plt.figure(figsize=(5, 4))
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", xticklabels=SIGNS, yticklabels=SIGNS)
    plt.title(title); plt.ylabel("true"); plt.xlabel("pred"); plt.tight_layout(); plt.show()
```

```python
rf_true, rf_pred = [], []
for tr, te in logo.split(X_tab, y_idx, groups):
    clf = RandomForestClassifier(n_estimators=300, random_state=0)
    clf.fit(X_tab[tr], y_idx[tr])
    p = clf.predict(X_tab[te])
    rf_true += list(y_idx[te]); rf_pred += list(p)
    print(groups[te][0], "acc:", round(accuracy_score(y_idx[te], p), 3))
report(rf_true, rf_pred, "Random Forest")
```

```python
class CNN(nn.Module):
    def __init__(self, c=162, n=6):
        super().__init__()
        self.net = nn.Sequential(
            nn.Conv1d(c, 64, 5, padding=2), nn.ReLU(), nn.MaxPool1d(2),
            nn.Conv1d(64, 128, 5, padding=2), nn.ReLU(), nn.AdaptiveAvgPool1d(1))
        self.fc = nn.Linear(128, n)
    def forward(self, x):
        return self.fc(self.net(x).squeeze(-1))

def train_eval(Xtr, ytr, Xte, epochs=60):
    mu = Xtr.mean((0, 2), keepdims=True); sd = Xtr.std((0, 2), keepdims=True) + 1e-6
    Xtr = torch.tensor((Xtr - mu) / sd, dtype=torch.float32)
    Xte = torch.tensor((Xte - mu) / sd, dtype=torch.float32)
    ytr = torch.tensor(ytr)
    m = CNN(); opt = torch.optim.Adam(m.parameters(), 1e-3); lossf = nn.CrossEntropyLoss()
    m.train()
    for _ in range(epochs):
        opt.zero_grad(); lossf(m(Xtr), ytr).backward(); opt.step()
    m.eval()
    with torch.no_grad():
        return m(Xte).argmax(1).numpy()
```

```python
cnn_true, cnn_pred = [], []
for tr, te in logo.split(X_seq, y_idx, groups):
    p = train_eval(X_seq[tr], y_idx[tr], X_seq[te])
    cnn_true += list(y_idx[te]); cnn_pred += list(p)
    print(groups[te][0], "acc:", round(accuracy_score(y_idx[te], p), 3))
report(cnn_true, cnn_pred, "1D-CNN")
```
