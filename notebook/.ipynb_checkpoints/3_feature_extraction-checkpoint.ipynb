# 3. Feature Extraction

```python
import numpy as np

SIGNS = ["Hello", "ThankYou", "Yes", "No", "Help", "Rest"]
d = np.load("processed.npz", allow_pickle=True)
X, y, groups = d["X"], d["y"], d["groups"]
```

```python
X_tab = np.concatenate([X.mean(axis=1), X.std(axis=1)], axis=1)   # RF: (N, 324)
X_seq = X.transpose(0, 2, 1)                                       # CNN: (N, 162, T)
y_idx = np.array([SIGNS.index(v) for v in y])
print("tabular:", X_tab.shape, "| sequence:", X_seq.shape)
```

```python
np.savez("features.npz", X_tab=X_tab, X_seq=X_seq, y_idx=y_idx, groups=groups)
print("saved features.npz")
```
