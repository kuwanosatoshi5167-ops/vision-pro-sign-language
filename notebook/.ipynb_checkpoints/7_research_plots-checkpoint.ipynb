# 7. Research-grade plots (per subject x per class)

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
from sklearn.decomposition import PCA

sns.set_theme(style="whitegrid", context="paper")
plt.rcParams.update({"figure.dpi": 120, "axes.titlesize": 11, "font.size": 10})

SIGNS = ["Hello", "ThankYou", "Yes", "No", "Help", "Rest"]
d = np.load("processed.npz", allow_pickle=True)
X, y, groups = d["X"], d["y"], d["groups"]          # X: (N, 90, 162) head-relative
coord_cols = list(d["coord_cols"])
SUBJECTS = sorted(set(groups))
T = X.shape[1]
SPAL = dict(zip(SUBJECTS, sns.color_palette("tab10", len(SUBJECTS))))
CPAL = dict(zip(SIGNS, sns.color_palette("Set2", len(SIGNS))))

def cidx(name): return coord_cols.index(name)
def chan(name): return X[:, :, cidx(name)]
JOINTS = ["wrist", "forearmWrist", "forearmArm",
          "thumbKnuckle", "thumbIntermediateBase", "thumbIntermediateTip", "thumbTip",
          "indexFingerMetacarpal", "indexFingerKnuckle", "indexFingerIntermediateBase", "indexFingerIntermediateTip", "indexFingerTip",
          "middleFingerMetacarpal", "middleFingerKnuckle", "middleFingerIntermediateBase", "middleFingerIntermediateTip", "middleFingerTip",
          "ringFingerMetacarpal", "ringFingerKnuckle", "ringFingerIntermediateBase", "ringFingerIntermediateTip", "ringFingerTip",
          "littleFingerMetacarpal", "littleFingerKnuckle", "littleFingerIntermediateBase", "littleFingerIntermediateTip", "littleFingerTip"]
print("ready:", X.shape)
```

### A. Spatial hand paths — dominant (right) index-finger trajectory in the head frame

```python
tx, ty = cidx("R_indexFingerTip_x"), cidx("R_indexFingerTip_y")
fig, axes = plt.subplots(2, 3, figsize=(13, 8), sharex=True, sharey=True)
for ax, sign in zip(axes.ravel(), SIGNS):
    for subj in SUBJECTS:
        idx = np.where((y == sign) & (groups == subj))[0]
        for i in idx:
            ax.plot(X[i, :, tx], X[i, :, ty], color=SPAL[subj], alpha=0.5, lw=0.9)
        m = X[idx].mean(0)
        ax.plot(m[:, tx], m[:, ty], color=SPAL[subj], lw=2.2)
        ax.scatter(m[0, tx], m[0, ty], color=SPAL[subj], marker="o", s=25, zorder=5)
    ax.set_title(sign); ax.set_xlabel("x (m)"); ax.set_ylabel("y (m)")
handles = [plt.Line2D([0], [0], color=SPAL[s], lw=2, label=s) for s in SUBJECTS]
fig.legend(handles=handles, loc="upper center", ncol=len(SUBJECTS), bbox_to_anchor=(0.5, 1.02))
fig.suptitle("Right index-fingertip path (bold = subject mean, dot = start)", y=0.98)
plt.tight_layout(); plt.show()
```

### B. Trajectory ribbons — mean ± σ of fingertip height across subjects

```python
fig, axes = plt.subplots(2, 3, figsize=(13, 7), sharex=True, sharey=True)
for ax, sign in zip(axes.ravel(), SIGNS):
    W = chan("R_indexFingerTip_y")[y == sign]
    mu, sd = W.mean(0), W.std(0)
    t = np.arange(T)
    ax.fill_between(t, mu - sd, mu + sd, color=CPAL[sign], alpha=0.3)
    ax.plot(t, mu, color=CPAL[sign], lw=2)
    for subj in SUBJECTS:
        sm = chan("R_indexFingerTip_y")[(y == sign) & (groups == subj)].mean(0)
        ax.plot(t, sm, color=SPAL[subj], lw=0.8, alpha=0.7)
    ax.set_title(sign); ax.set_xlabel("frame"); ax.set_ylabel("y (m)")
fig.suptitle("R index-fingertip height: population mean ± σ (band) with per-subject means (thin)", y=0.99)
plt.tight_layout(); plt.show()
```

### C. Range-of-motion heatmap — per-joint movement magnitude by sign

```python
def joint_rom(sign, hand):
    rom = []
    W = X[y == sign]
    for j in JOINTS:
        p = W[:, :, cidx(f"{hand}_{j}_x"):cidx(f"{hand}_{j}_x") + 3]   # (n, T, 3)
        rom.append(np.linalg.norm(p.std(axis=1), axis=1).mean())        # mean over takes of positional std magnitude
    return rom

fig, axes = plt.subplots(2, 1, figsize=(13, 7))
for ax, hand, name in zip(axes, ("R", "L"), ("Right (dominant)", "Left")):
    M = np.array([joint_rom(s, hand) for s in SIGNS])
    sns.heatmap(M, ax=ax, cmap="viridis", xticklabels=JOINTS, yticklabels=SIGNS,
                cbar_kws={"label": "movement (m)"})
    ax.set_title(f"{name} hand — per-joint range of motion")
    ax.set_xticklabels(JOINTS, rotation=90, fontsize=7)
plt.tight_layout(); plt.show()
```

### D. Per-class 3D pose overlay — subjects coloured (frame 45)

```python
FINGERS = [["thumbKnuckle", "thumbIntermediateBase", "thumbIntermediateTip", "thumbTip"],
           ["indexFingerMetacarpal", "indexFingerKnuckle", "indexFingerIntermediateBase", "indexFingerIntermediateTip", "indexFingerTip"],
           ["middleFingerMetacarpal", "middleFingerKnuckle", "middleFingerIntermediateBase", "middleFingerIntermediateTip", "middleFingerTip"],
           ["ringFingerMetacarpal", "ringFingerKnuckle", "ringFingerIntermediateBase", "ringFingerIntermediateTip", "ringFingerTip"],
           ["littleFingerMetacarpal", "littleFingerKnuckle", "littleFingerIntermediateBase", "littleFingerIntermediateTip", "littleFingerTip"]]
BONES = []
for ch in FINGERS:
    BONES.append(("wrist", ch[0])); BONES += list(zip(ch[:-1], ch[1:]))
BONES += [("wrist", "forearmWrist"), ("forearmWrist", "forearmArm")]
FRAME = 45
lo = X.reshape(-1, 3).min(0); hi = X.reshape(-1, 3).max(0)

def draw(ax, w, col):
    for hand in ("L", "R"):
        wr = w[FRAME, cidx(f"{hand}_wrist_x"):cidx(f"{hand}_wrist_x") + 3]
        if np.abs(wr).sum() < 1e-6: continue
        P = {j: w[FRAME, cidx(f"{hand}_{j}_x"):cidx(f"{hand}_{j}_x") + 3] for j in JOINTS}
        for a, b in BONES:
            ax.plot(*[[P[a][k], P[b][k]] for k in range(3)], color=col, lw=1.1)

fig = plt.figure(figsize=(14, 8))
for n, sign in enumerate(SIGNS):
    ax = fig.add_subplot(2, 3, n + 1, projection="3d")
    for subj in SUBJECTS:
        idx = np.where((y == sign) & (groups == subj))[0]
        draw(ax, X[idx[0]], SPAL[subj])
    ax.set_xlim(lo[0], hi[0]); ax.set_ylim(lo[1], hi[1]); ax.set_zlim(lo[2], hi[2])
    ax.set_xticks([]); ax.set_yticks([]); ax.set_zticks([])
    ax.view_init(elev=15, azim=-70); ax.set_title(sign)
fig.legend(handles=[plt.Line2D([0], [0], color=SPAL[s], lw=2, label=s) for s in SUBJECTS],
           loc="upper center", ncol=len(SUBJECTS))
fig.suptitle("Hand pose per sign, subjects overlaid", y=0.98)
plt.tight_layout(); plt.show()
```

### E. State-space trajectory — class mean paths through pose PCA

```python
pca = PCA(n_components=2, random_state=0).fit(X.reshape(-1, X.shape[2]))
plt.figure(figsize=(7.5, 6.5))
for sign in SIGNS:
    m = X[y == sign].mean(0)                     # (T, 162)
    z = pca.transform(m)                          # (T, 2)
    plt.plot(z[:, 0], z[:, 1], color=CPAL[sign], lw=2, label=sign)
    plt.scatter(z[0, 0], z[0, 1], color=CPAL[sign], marker="o", s=40, zorder=5)
    plt.scatter(z[-1, 0], z[-1, 1], color=CPAL[sign], marker="s", s=40, zorder=5)
plt.xlabel("PC1"); plt.ylabel("PC2")
plt.title("Class mean trajectory in pose PCA space (o = start, square = end)")
plt.legend(); plt.tight_layout(); plt.show()
```

### F. Inter-subject consistency — mean pairwise trajectory correlation per sign

```python
def consistency(sign):
    subj_means = [chan("R_indexFingerTip_y")[(y == sign) & (groups == s)].mean(0) for s in SUBJECTS]
    C = np.corrcoef(subj_means)
    iu = np.triu_indices(len(SUBJECTS), 1)
    return C[iu].mean()

vals = [consistency(s) for s in SIGNS]
plt.figure(figsize=(7, 4))
sns.barplot(x=SIGNS, y=vals, palette=[CPAL[s] for s in SIGNS])
plt.ylabel("mean pairwise corr (across subjects)"); plt.ylim(0, 1)
plt.title("How consistently is each sign performed across subjects?")
plt.tight_layout(); plt.show()
```
