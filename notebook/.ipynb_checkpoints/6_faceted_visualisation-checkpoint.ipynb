# 6. Faceted Visualisation — per subject × per class

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

SIGNS = ["Hello", "ThankYou", "Yes", "No", "Help", "Rest"]
d = np.load("processed.npz", allow_pickle=True)
X, y, groups = d["X"], d["y"], d["groups"]          # X: (N, 90, 162) head-relative
coord_cols = list(d["coord_cols"])
SUBJECTS = sorted(set(groups))
FRAME = 45
print("subjects:", SUBJECTS, "| signs:", SIGNS)
```

```python
FINGERS = {
    "thumb":  ["thumbKnuckle", "thumbIntermediateBase", "thumbIntermediateTip", "thumbTip"],
    "index":  ["indexFingerMetacarpal", "indexFingerKnuckle", "indexFingerIntermediateBase", "indexFingerIntermediateTip", "indexFingerTip"],
    "middle": ["middleFingerMetacarpal", "middleFingerKnuckle", "middleFingerIntermediateBase", "middleFingerIntermediateTip", "middleFingerTip"],
    "ring":   ["ringFingerMetacarpal", "ringFingerKnuckle", "ringFingerIntermediateBase", "ringFingerIntermediateTip", "ringFingerTip"],
    "little": ["littleFingerMetacarpal", "littleFingerKnuckle", "littleFingerIntermediateBase", "littleFingerIntermediateTip", "littleFingerTip"],
}
BONES = []
for chain in FINGERS.values():
    BONES.append(("wrist", chain[0])); BONES += list(zip(chain[:-1], chain[1:]))
BONES += [("wrist", "forearmWrist"), ("forearmWrist", "forearmArm")]
ALL_JOINTS = ["wrist", "forearmWrist", "forearmArm"] + [j for ch in FINGERS.values() for j in ch]

def joint_xyz(window, hand, joint):
    i = coord_cols.index(f"{hand}_{joint}_x"); return window[:, i:i+3]

def hand_present(window, hand):
    i = coord_cols.index(f"{hand}_wrist_x"); return np.abs(window[:, i:i+3]).sum() > 1e-6

def rep_window(sign, subject):
    idx = np.where((y == sign) & (groups == subject))[0]
    return X[idx[0]] if len(idx) else None

def draw_hands(ax, window, frame, colors=("tab:blue", "tab:red")):
    for hand, col in zip(("L", "R"), colors):
        if not hand_present(window, hand): continue
        P = {j: joint_xyz(window, hand, j)[frame] for j in ALL_JOINTS}
        for a, b in BONES:
            xa, xb = P[a], P[b]
            ax.plot([xa[0], xb[0]], [xa[1], xb[1]], [xa[2], xb[2]], color=col, lw=1.2)
        pts = np.array(list(P.values()))
        ax.scatter(pts[:, 0], pts[:, 1], pts[:, 2], color=col, s=5)
```

### Pose grid — rows = subjects, columns = signs (blue = left hand, red = right)

```python
lo = X.reshape(-1, 3).min(0); hi = X.reshape(-1, 3).max(0)
nR, nC = len(SUBJECTS), len(SIGNS)
fig = plt.figure(figsize=(3 * nC, 3 * nR))
for r, subj in enumerate(SUBJECTS):
    for c, sign in enumerate(SIGNS):
        ax = fig.add_subplot(nR, nC, r * nC + c + 1, projection="3d")
        w = rep_window(sign, subj)
        if w is not None:
            draw_hands(ax, w, FRAME)
        ax.set_xlim(lo[0], hi[0]); ax.set_ylim(lo[1], hi[1]); ax.set_zlim(lo[2], hi[2])
        ax.set_xticks([]); ax.set_yticks([]); ax.set_zticks([])
        ax.view_init(elev=15, azim=-70)
        if r == 0: ax.set_title(sign, fontsize=11)
        if c == 0: ax.text2D(-0.15, 0.5, subj, transform=ax.transAxes, fontsize=11, rotation=90, va="center")
fig.suptitle(f"Hand pose per subject x sign (head-relative, frame {FRAME})", fontsize=13)
plt.tight_layout(); plt.show()
```

### Motion-speed grid — rows = subjects, columns = signs (faint = each take, bold = mean)

```python
def frame_speed(window):
    pts = window.reshape(window.shape[0], -1, 3)
    return np.linalg.norm(np.diff(pts, axis=0), axis=2).mean(axis=1)

fig, axes = plt.subplots(nR, nC, figsize=(2.6 * nC, 2.2 * nR), sharex=True, sharey=True)
for r, subj in enumerate(SUBJECTS):
    for c, sign in enumerate(SIGNS):
        ax = axes[r, c]
        idx = np.where((y == sign) & (groups == subj))[0]
        curves = np.stack([frame_speed(X[i]) for i in idx])
        for cu in curves:
            ax.plot(cu, color="gray", alpha=0.35, lw=0.8)
        ax.plot(curves.mean(0), color="C3", lw=1.8)
        if r == 0: ax.set_title(sign, fontsize=10)
        if c == 0: ax.set_ylabel(subj, fontsize=10)
fig.suptitle("Mean joint speed over time, per subject x sign", fontsize=13)
plt.tight_layout(); plt.show()
```

### Fingertip trajectory grid — R index-fingertip y over time (all takes)

```python
ch = coord_cols.index("R_indexFingerTip_y")
fig, axes = plt.subplots(nR, nC, figsize=(2.6 * nC, 2.2 * nR), sharex=True, sharey=True)
for r, subj in enumerate(SUBJECTS):
    for c, sign in enumerate(SIGNS):
        ax = axes[r, c]
        idx = np.where((y == sign) & (groups == subj))[0]
        for i in idx:
            ax.plot(X[i, :, ch], color="C0", alpha=0.5, lw=0.8)
        if r == 0: ax.set_title(sign, fontsize=10)
        if c == 0: ax.set_ylabel(subj, fontsize=10)
fig.suptitle("R index-fingertip y (head-relative), per subject x sign", fontsize=13)
plt.tight_layout(); plt.show()
```
