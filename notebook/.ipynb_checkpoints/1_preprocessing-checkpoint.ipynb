# 1. Data Preprocessing

```python
import glob, os
import numpy as np
import pandas as pd

DATA_DIR = "../Data"
CACHE = "processed.npz"
SIGNS = ["Hello", "ThankYou", "Yes", "No", "Help", "Rest"]
HEAD_POS = ["head_x", "head_y", "head_z"]
HEAD_QUAT = ["head_qx", "head_qy", "head_qz", "head_qw"]
T = 90
```

```python
def load_windows(data_dir):
    coord_cols = None
    rows = []
    for f in sorted(glob.glob(os.path.join(data_dir, "*.csv"))):
        df = pd.read_csv(f)
        df = df[(df.is_warmup == 0) & (df.kept == 1)]
        if coord_cols is None:
            coord_cols = [c for c in df.columns
                          if (c.startswith("L_") or c.startswith("R_")) and c[-2:] in ("_x", "_y", "_z")]
        for (sid, wid), g in df.groupby(["subject_id", "window_id"]):
            g = g.sort_values("frame_index")
            joints = g[coord_cols].to_numpy(dtype=float)
            head_p = g[HEAD_POS].to_numpy(dtype=float)
            head_q = g[HEAD_QUAT].to_numpy(dtype=float)
            rows.append((sid, g.sign_label.iloc[0], joints, head_p, head_q))
    return rows, coord_cols

windows, coord_cols = load_windows(DATA_DIR)
lengths = np.array([w[2].shape[0] for w in windows])
print("windows:", len(windows), "| coord columns:", len(coord_cols),
      "| frames min/med/max:", lengths.min(), int(np.median(lengths)), lengths.max())
```

```python
def quat_to_R(q):                      # (x, y, z, w) head->world rotation matrix
    x, y, z, w = q
    return np.array([
        [1 - 2*(y*y + z*z), 2*(x*y - w*z),     2*(x*z + w*y)],
        [2*(x*y + w*z),     1 - 2*(x*x + z*z), 2*(y*z - w*x)],
        [2*(x*z - w*y),     2*(y*z + w*x),     1 - 2*(x*x + y*y)]])

def to_head_frame(joints, head_p, head_q):
    F = joints.shape[0]
    pts = joints.reshape(F, -1, 3)                     # (F, 54, 3) world
    out = np.empty_like(pts)
    for t in range(F):
        R = quat_to_R(head_q[t])                       # head->world
        out[t] = (pts[t] - head_p[t]) @ R              # world->head
    return out.reshape(F, -1)

def fix_len(a, T):
    n = a.shape[0]
    if n >= T:
        return a[:T]
    return np.vstack([a, np.repeat(a[-1:], T - n, axis=0)])
```

```python
X = np.stack([fix_len(np.nan_to_num(to_head_frame(j, hp, hq), nan=0.0), T)
              for (_, _, j, hp, hq) in windows])
y = np.array([s for (_, s, _, _, _) in windows])
groups = np.array([sid for (sid, _, _, _, _) in windows])
print("X:", X.shape, "| subjects:", sorted(set(groups)))
```

```python
np.savez(CACHE, X=X, y=y, groups=groups, lengths=lengths, coord_cols=np.array(coord_cols))
print("saved", CACHE)
```
