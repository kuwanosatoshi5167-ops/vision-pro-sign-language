# 2. Data Visualisation

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

SIGNS = ["Hello", "ThankYou", "Yes", "No", "Help", "Rest"]
d = np.load("processed.npz", allow_pickle=True)
X, y, groups, lengths = d["X"], d["y"], d["groups"], d["lengths"]
coord_cols = list(d["coord_cols"])
```

```python
meta = pd.DataFrame({"subject": groups, "sign": y})
fig, ax = plt.subplots(1, 2, figsize=(13, 4))
sns.countplot(data=meta, x="sign", hue="subject", order=SIGNS, ax=ax[0])
ax[0].set_title("Windows per sign per subject"); ax[0].tick_params(axis="x", rotation=30)
ax[1].hist(lengths, bins=range(lengths.min(), lengths.max() + 2))
ax[1].set_title("Frames per window"); ax[1].set_xlabel("frames")
plt.tight_layout(); plt.show()
```

```python
ch = coord_cols.index("R_indexFingerTip_y")
plt.figure(figsize=(9, 4))
for s in SIGNS:
    i = np.where(y == s)[0][0]
    plt.plot(X[i, :, ch], label=s)
plt.title("R index-fingertip y (head-relative) over time"); plt.xlabel("frame"); plt.legend()
plt.tight_layout(); plt.show()
```

```python
flat = X.reshape(len(X), -1)
pcs = PCA(n_components=2, random_state=0).fit_transform(StandardScaler().fit_transform(flat))
plt.figure(figsize=(6, 5))
sns.scatterplot(x=pcs[:, 0], y=pcs[:, 1], hue=y, hue_order=SIGNS, s=30)
plt.title("PCA of head-relative windows"); plt.tight_layout(); plt.show()
```
